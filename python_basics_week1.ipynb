```python
def analyze_string(text):

    vowels = "aeiouAEIOU"

    result = {}

    result["length"] = len(text)
    result["uppercase"] = text.upper()
    result["lowercase"] = text.lower()
    result["title"] = text.title()
    result["word_count"] = len(text.split())

    vowel_count = 0
    consonant_count = 0
    digit_count = 0
    space_count = 0
    special_count = 0

    for char in text:

        if char.isalpha():

            if char in vowels:
                vowel_count += 1
            else:
                consonant_count += 1

        elif char.isdigit():
            digit_count += 1

        elif char.isspace():
            space_count += 1

        else:
            special_count += 1

    result["vowels"] = vowel_count
    result["consonants"] = consonant_count
    result["digits"] = digit_count
    result["spaces"] = space_count
    result["special_characters"] = special_count

    return result
```

```python
text = "Python is the best language for Data Analysis"

analyze_string(text)
```

```python
def string(text):

  vowels = "aeiouAEIOU"

  result = {}

  vowels_count = 0
  consonent_count = 0

  for char in text:

    if char in vowels:
      vowels_count += 1

    else:
      consonent_count += 1

  result["vowels"] = vowels_count
  result["consonants"] = consonent_count

  return result
```
