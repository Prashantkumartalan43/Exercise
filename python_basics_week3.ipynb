```python
class Product:
  def __init__(self, name, price):

    self.name = name
    self.price = price

  def discounted_price(self, discount):

    return(self.price * (1 - discount / 100))
```

```python
def city_report(df):
  report = (df.groupby('City').agg(
      Orders = ('Customer', 'count'),
      Revenue = ('Sales', 'sum'),
      Profit = ('Profit', 'sum'),
      Average_sales = ('Sales', 'mean'),
      Highest_sale = ('Sales', 'max'),
      Lowest_sale = ('Sales', 'min')
      ).round(2)
  .reset_index()
    )
  return report
```

```python
report = city_report(df)

print(report)
```

```python
def category_kpi(df, group_col):
  report = (
        df.groupby(group_col)
        .agg(
            Orders=("Customer", "count"),
            Revenue=("Sales", "sum"),
            Profit=("Profit", "sum"),
            Average_Sales=("Sales", "mean"),
            Highest_Sale=("Sales", "max"),
            Lowest_Sale=("Sales", "min")
        )
        .round(2)
        .reset_index()
    )
  return report
```

```python
city_report = category_kpi(df, "City")

print(city_report)
```

```python
category_report = category_kpi(df, "Category")

print(category_report)
```

```python
def category_kpi(df, group_col, sort_by="Revenue", ascending=False):

    report = (
        df.groupby(group_col)
        .agg(
            Orders=("Customer", "count"),
            Revenue=("Sales", "sum"),
            Profit=("Profit", "sum"),
            Average_Sales=("Sales", "mean"),
            Highest_Sale=("Sales", "max"),
            Lowest_Sale=("Sales", "min")
        )
        .round(2)
        .sort_values(by=sort_by, ascending=ascending)
        .reset_index()
    )

    return report
```

```python
category_kpi(
    df,
    "City",
    sort_by="Profit"
)
```

```python
category_kpi(
    df,
    "Category",
    sort_by="Orders"
)
```

```python
import pandas as pd

data = {
    "Customer":["Rahul","Amit","Priya","Neha","Vikas","Riya"],
    "Sales":[50000,30000,70000,25000,60000,45000],
    "Profit":[8000,3000,12000,-2000,9000,7000],
    "Category":[
        "Electronics",
        "Furniture",
        "Electronics",
        "Furniture",
        "Office Supplies",
        "Electronics"
    ]
}

df = pd.DataFrame(data)
```

```python
def sales_category(sales):
    if sales >= 60000:
        return "High"
    elif sales >= 40000:
        return "Medium"
    else:
        return "Low"

df["Sales_Category"] = df["Sales"].apply(sales_category)

print(df)
```

```python
def profit_status(profit):
    if profit < 0:
        return "Loss"
    else:
        return "Profit"

df["Profit_Status"] = df["Profit"].apply(profit_status)

print(df)
```

```python
def customer_tier(sales):
    if sales >= 60000:
        return "Gold"
    elif sales >= 40000:
        return "Silver"
    else:
        return "Bronze"

df["Customer_Tier"] = df["Sales"].apply(customer_tier)

print(df)
```

```python
def classify_sales(df):

    df = df.copy()

    def sales_category(sales):
        if sales >= 60000:
            return "High"
        elif sales >= 40000:
            return "Medium"
        else:
            return "Low"

    def profit_status(profit):
        if profit < 0:
            return "Loss"
        else:
            return "Profit"

    def customer_tier(sales):
        if sales >= 60000:
            return "Gold"
        elif sales >= 40000:
            return "Silver"
        else:
            return "Bronze"

    df["Sales_Category"] = df["Sales"].apply(sales_category)
    df["Profit_Status"] = df["Profit"].apply(profit_status)
    df["Customer_Tier"] = df["Sales"].apply(customer_tier)

    return df
```

```python
result = classify_sales(df)

print(result)
```
